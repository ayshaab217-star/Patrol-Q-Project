```python
import pandas as pd
import numpy as np

df = pd.read_csv("PatrolIQ_cleaned.csv")

print(df.shape)
df.head()
```

```python
print(df.columns.tolist())
```

```python
# ==========================================
# FEATURE ENGINEERING 1 - TEMPORAL FEATURES
# ==========================================

# Convert date to datetime
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Create temporal features
df["hour"] = df["date"].dt.hour
df["day_of_week"] = df["date"].dt.day_name()
df["month"] = df["date"].dt.month

# Weekend flag
df["is_weekend"] = df["day_of_week"].isin(
    ["Saturday", "Sunday"]
)

# Season
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

df["season"] = df["month"].apply(get_season)

print(df[
    ["date", "hour", "day_of_week", "month",
     "is_weekend", "season"]
].head())
```

```python
# ==========================================
# FEATURE ENGINEERING 2 - CRIME SEVERITY
# ==========================================

severity_map = {
    "NON-CRIMINAL": 1,
    "THEFT": 2,
    "DECEPTIVE PRACTICE": 2,
    "NARCOTICS": 3,
    "BURGLARY": 3,
    "CRIMINAL DAMAGE": 3,
    "CRIMINAL TRESPASS": 3,
    "MOTOR VEHICLE THEFT": 3,
    "ASSAULT": 4,
    "BATTERY": 4,
    "ROBBERY": 4,
    "WEAPONS VIOLATION": 4,
    "HOMICIDE": 5,
    "KIDNAPPING": 5,
    "CRIMINAL SEXUAL ASSAULT": 5,
    "HUMAN TRAFFICKING": 5
}

df["crime_severity_score"] = (
    df["primary_type"]
    .map(severity_map)
    .fillna(2)
)

print(df[["primary_type", "crime_severity_score"]].head())
```

```python
# ==========================================
# FEATURE ENGINEERING 3 - GEOGRAPHIC FEATURES
# ==========================================

# Remove rows without coordinates
df = df.dropna(subset=["latitude", "longitude"]).copy()

# Create geographic grid
df["lat_zone"] = df["latitude"].round(2)
df["lon_zone"] = df["longitude"].round(2)

print(df[[
    "latitude",
    "longitude",
    "lat_zone",
    "lon_zone"
]].head())

print("\nShape after geographic cleaning:", df.shape)
```

```python
# ==========================================
# FEATURE ENGINEERING 4 - CATEGORICAL ENCODING
# ==========================================

# Frequency encode crime type
crime_freq = df["primary_type"].value_counts(normalize=True)
df["crime_type_freq"] = df["primary_type"].map(crime_freq)

# Frequency encode location
location_freq = df["location_description"].value_counts(normalize=True)
df["location_freq"] = df["location_description"].map(location_freq)

print(df[[
    "primary_type",
    "crime_type_freq",
    "location_description",
    "location_freq"
]].head())
```

```python
# ==========================================
# FEATURE ENGINEERING 5 - ML FEATURES
# ==========================================

features = [
    "latitude",
    "longitude",
    "hour",
    "month",
    "is_weekend",
    "crime_severity_score",
    "crime_type_freq",
    "location_freq",
    "district",
    "community_area"
]

X = df[features].copy()

print("ML Feature Shape:", X.shape)
print("\nFeatures:")
print(X.columns.tolist())

print("\nMissing Values:")
print(X.isna().sum())
```

```python
# ==========================================
# FEATURE ENGINEERING 6 - MISSING VALUES
# ==========================================

X["district"] = X["district"].fillna(X["district"].median())
X["community_area"] = X["community_area"].fillna(
    X["community_area"].median()
)

print("Remaining Missing Values:")
print(X.isna().sum())
```

```python
# ==========================================
# FEATURE ENGINEERING 7 - FEATURE SCALING
# ==========================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled Feature Shape:", X_scaled.shape)
print("\nFirst 5 rows:")
print(X_scaled[:5])
```

```python
# ==========================================
# FEATURE ENGINEERING 8 - SAVE ML DATA
# ==========================================

X.to_csv("PatrolIQ_ML_Features.csv", index=False)

print("ML-ready dataset saved successfully!")
print("Shape:", X.shape)
```
