```python
import pandas as pd

file_path = r"D:\PatrolQ project\ijzp-q8t2.csv"

df = pd.read_csv(file_path, low_memory=False)

print("Dataset loaded successfully!")
print(df.shape)
```

```python
# ==========================================
# CASE NUMBER CLEANING
# ==========================================

# Convert to string
df["case_number"] = df["case_number"].astype("string")

# Remove leading/trailing spaces
df["case_number"] = df["case_number"].str.strip()

# Convert to uppercase
df["case_number"] = df["case_number"].str.upper()

# Check missing values
print("Missing case numbers:")
print(df["case_number"].isna().sum())

# Check empty values
print("\nEmpty case numbers:")
print((df["case_number"] == "").sum())

# Check duplicates
print("\nDuplicate case numbers:")
print(df["case_number"].duplicated().sum())

# Check data type
print("\nData type:")
print(df["case_number"].dtype)

# Check unique values
print("\nUnique case numbers:")
print(df["case_number"].nunique())

# Show sample
print("\nFirst 10 case numbers:")
print(df["case_number"].head(10))
```

```python
# Check duplicate case numbers

duplicate_cases = df[
    df["case_number"].duplicated(keep=False)
].sort_values("case_number")

print("Number of rows with duplicate case numbers:")
print(len(duplicate_cases))

print("\nDuplicate case numbers:")
print(
    duplicate_cases[
        ["id", "case_number", "date", "primary_type", "description"]
    ].to_string(index=False)
)
```

```python
# ==========================================
# DATE CLEANING
# ==========================================

# Convert date to datetime
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)

# Check data type
print("Data type:")
print(df["date"].dtype)

# Check missing/invalid dates
print("\nMissing or invalid dates:")
print(df["date"].isna().sum())

# Check minimum date
print("\nEarliest date:")
print(df["date"].min())

# Check maximum date
print("\nLatest date:")
print(df["date"].max())

# Check duplicate date values
print("\nUnique dates:")
print(df["date"].nunique())

# Show first 10 cleaned dates
print("\nFirst 10 dates:")
print(df["date"].head(10))
```

```python
# ==========================================
# BLOCK COLUMN CLEANING
# ==========================================

# Convert to string
df["block"] = df["block"].astype("string")

# Remove leading and trailing spaces
df["block"] = df["block"].str.strip()

# Convert to uppercase
df["block"] = df["block"].str.upper()

# Check missing values
print("Missing block values:")
print(df["block"].isna().sum())

# Check empty values
print("\nEmpty block values:")
print((df["block"] == "").sum())

# Check unique values
print("\nUnique block values:")
print(df["block"].nunique())

# Check data type
print("\nData type:")
print(df["block"].dtype)

# Display first 10
print("\nFirst 10 block values:")
print(df["block"].head(10))
```

```python
# ==========================================
# IUCR COLUMN CLEANING
# ==========================================

# Convert to string
df["iucr"] = df["iucr"].astype("string")

# Remove leading/trailing spaces
df["iucr"] = df["iucr"].str.strip()

# Convert to uppercase
df["iucr"] = df["iucr"].str.upper()

# Check missing values
print("Missing IUCR values:")
print(df["iucr"].isna().sum())

# Check empty values
print("\nEmpty IUCR values:")
print((df["iucr"] == "").sum())

# Check unique values
print("\nUnique IUCR codes:")
print(df["iucr"].nunique())

# Check data type
print("\nData type:")
print(df["iucr"].dtype)

# Show most common IUCR codes
print("\nTop 10 IUCR codes:")
print(df["iucr"].value_counts().head(10))

# Show first 10
print("\nFirst 10 IUCR values:")
print(df["iucr"].head(10))
```

```python
# ==========================================
# PRIMARY TYPE COLUMN CLEANING
# ==========================================

# Convert to string
df["primary_type"] = df["primary_type"].astype("string")

# Remove leading/trailing spaces
df["primary_type"] = df["primary_type"].str.strip()

# Convert to uppercase
df["primary_type"] = df["primary_type"].str.upper()

# Check missing values
print("Missing primary type values:")
print(df["primary_type"].isna().sum())

# Check empty values
print("\nEmpty primary type values:")
print((df["primary_type"] == "").sum())

# Check unique crime types
print("\nUnique crime types:")
print(df["primary_type"].nunique())

# Display all crime types
print("\nCrime types:")
print(sorted(df["primary_type"].dropna().unique()))

# Show crime frequency
print("\nTop 15 crime types:")
print(df["primary_type"].value_counts().head(15))

# Check data type
print("\nData type:")
print(df["primary_type"].dtype)
```

```python
# ==========================================
# DESCRIPTION COLUMN CLEANING
# ==========================================

# Convert to string
df["description"] = df["description"].astype("string")

# Remove leading/trailing spaces
df["description"] = df["description"].str.strip()

# Convert to uppercase
df["description"] = df["description"].str.upper()

# Replace multiple spaces with a single space
df["description"] = df["description"].str.replace(
    r"\s+",
    " ",
    regex=True
)

# Check missing values
print("Missing description values:")
print(df["description"].isna().sum())

# Check empty values
print("\nEmpty description values:")
print((df["description"] == "").sum())

# Check unique descriptions
print("\nUnique descriptions:")
print(df["description"].nunique())

# Check data type
print("\nData type:")
print(df["description"].dtype)

# Show most common descriptions
print("\nTop 15 descriptions:")
print(df["description"].value_counts().head(15))

# Show first 10
print("\nFirst 10 descriptions:")
print(df["description"].head(10))
```

```python
# ==========================================
# COMMUNITY AREA FINAL CLEANING
# ==========================================

# Convert to numeric and nullable integer
df["community_area"] = pd.to_numeric(
    df["community_area"],
    errors="coerce"
).astype("Int64")

# Check missing values
print("Missing community area values:")
print(df["community_area"].isna().sum())

# Check invalid values
print("\nInvalid community area values:")
print(
    (
        (df["community_area"] < 1) |
        (df["community_area"] > 77)
    ).sum()
)

# Check unique values
print("\nUnique community areas:")
print(df["community_area"].nunique())

# Check minimum
print("\nMinimum community area:")
print(df["community_area"].min())

# Check maximum
print("\nMaximum community area:")
print(df["community_area"].max())

# Check data type
print("\nData type:")
print(df["community_area"].dtype)
```
