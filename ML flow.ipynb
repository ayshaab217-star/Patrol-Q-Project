```python
# ==========================================
# PHASE 11 - MLFLOW
# ==========================================

import mlflow

mlflow.set_experiment("PatrolIQ_Clustering")

print("MLflow experiment ready!")
```

```python
with mlflow.start_run(run_name="KMeans_Clustering"):

    mlflow.log_param("algorithm", "K-Means")
    mlflow.log_param("n_clusters", 5)
    mlflow.log_param("random_state", 42)

    mlflow.log_metric("silhouette_score", 0.139)

print("K-Means experiment logged!")
```

```python
with mlflow.start_run(run_name="DBSCAN_Clustering"):

    mlflow.log_param("algorithm", "DBSCAN")
    mlflow.log_param("eps", 1.5)
    mlflow.log_param("min_samples", 10)

    mlflow.log_metric("silhouette_score", 0.026)

print("DBSCAN experiment logged!")
```

```python
# ==========================================
# CHECK MLFLOW EXPERIMENT
# ==========================================

experiment = mlflow.get_experiment_by_name(
    "PatrolIQ_Clustering"
)

print("Experiment ID:", experiment.experiment_id)
print("Experiment Name:", experiment.name)

runs = mlflow.search_runs(
    experiment_ids=[experiment.experiment_id]
)

print("\nMLflow Runs:")
print(
    runs[
        ["run_id", "status", "params.algorithm", "metrics.silhouette_score"]
    ]
)
```
